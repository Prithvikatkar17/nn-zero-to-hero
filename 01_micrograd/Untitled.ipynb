```python
class value:

    def __init__(self,data ,_children=() ,_op=''):
        self.data = data 
        self._prev = set(_children)
        self._op = _op
        
    def __repr__(self):
        return f"Value(data = {self.data})"

    def __add__(self ,other):
        out = value(self.data + other.data,(self,other) ,'+') # a.__add__(b)
        return out 
        
    def __mul__(self ,other):
        out = value(self.data * other.data,(self ,other) ,'*') # a.__add__(b)
        return out 
        

a = value(3)
b = value(-4)
c = value(3.0)
a +b
d = a*b + c
```
